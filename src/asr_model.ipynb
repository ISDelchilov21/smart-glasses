```python
import librosa
from datasets import Audio
MAX_DURATION_IN_SECONDS = 20.0

def is_audio_lenght_in_range(input_length):
    return input_length <= MAX_DURATION_IN_SECONDS


minds = minds.cast_column("audio", Audio(sampling_rate=16_000))

new_column = [librosa.get_duration(y=x["array"], sr=x["sampling_rate"]) for x in minds["audio"]]

minds = minds.add_column("duration", new_column)

minds = minds.filter(is_audio_lenght_in_range, input_columns=["duration"])

minds = minds.remove_columns("duration")
minds
```

```python
from transformers import WhisperFeatureExtractor

feature_extractor = WhisperFeatureExtractor.from_pretrained("openai/whisper-small")

def prepare_dataset(example):

    audio = example["audio"]
    if audio["sampling_rate"] != 16000:
        audio_arr = librosa.resample(audio["array"], orig_sr = audio["sampling_rate"], target_sr=16000 )
        audio = {"array": audio_arr, "sampling_rate": 16000}
    features = feature_extractor(audio["array"], sampling_rate = audio["sampling_rate"], padding = True)
    return features

minds = minds.map(prepare_dataset)
minds
```

```python
from transformers import pipeline
import gradio as gr
import numpy as np
import librosa
TARGET_SR = 16_000
model_id = "openai/whisper-small"
pipe = pipeline("automatic-speech-recognition", model=model_id)

def transcribe_stream(state, chunk):
    if chunk is None:
        return state, (state[2] if state else "")
    sr, data = chunk
    data = data.astype(np.float32)
    if data.ndim > 1:
        data = data.mean(axis=1)
    if np.abs(data).max() > 1.0:
        data /= 32768.0
    if sr != TARGET_SR:
        data = librosa.resample(data, orig_sr=sr, target_sr=TARGET_SR)
        
    audio = data if state is None else np.concatenate([state[1], data])
    text = pipe({"raw": audio, "sampling_rate": 16_000},
                generate_kwargs={"task": "transcribe", "language": "english"})["text"]
    return (sr, audio, text), text

live = gr.Blocks()

live = gr.Interface(
    fn=transcribe_stream,
    inputs=["state", gr.Audio(sources="microphone", type="numpy", streaming=True)],
    outputs=["state", "text"],
    live=True,
)
live.launch(debug=True)
```

```python
from threading import Thread, Event
from transformers import AutoModelForRNNT, AutoProcessor, TextIteratorStreamer
import queue
import traceback
import numpy as np
import sounddevice as sd

model_id = "nvidia/nemotron-speech-streaming-en-0.6b"
processor = AutoProcessor.from_pretrained(model_id)
# load fully into RAM; device_map="auto" offloaded weights to disk and crashed the kernel
model = AutoModelForRNNT.from_pretrained(model_id).to("cpu")

processor.set_num_lookahead_tokens(6)
print(f"Streaming latency: {processor.streaming_latency_ms} ms")

sampling_rate = processor.feature_extractor.sampling_rate
MIC = "Headset CMF WASAPI"

mic_q = queue.Queue()
buffer = np.zeros(0, dtype=np.float32)
stop = Event()


# the microphone puts audio into the queue
def mic_callback(indata, frames, time, status):
    mic_q.put(indata[:, 0].copy())


# move audio from the queue into the buffer until it's long enough
def wait_for_samples(n):
    global buffer
    while len(buffer) < n:
        if stop.is_set():
            raise StopIteration
        try:
            buffer = np.concatenate([buffer, mic_q.get(timeout=0.5)])
        except queue.Empty:
            continue


def input_features_generator(first_chunk_inputs):
    yield first_chunk_inputs.input_features[:, : processor.num_mel_frames_first_audio_chunk, :]

    mel_frame_idx = processor.num_mel_frames_first_audio_chunk
    hop_length = processor.feature_extractor.hop_length
    n_fft = processor.feature_extractor.n_fft

    start_idx = mel_frame_idx * hop_length - n_fft // 2
    while True:  # a microphone never ends, so loop until stopped
        end_idx = start_idx + processor.num_samples_per_audio_chunk
        try:
            wait_for_samples(end_idx)
        except StopIteration:
            return
        inputs = processor(
            buffer[start_idx:end_idx],
            sampling_rate=sampling_rate,
            is_streaming=True,
            is_first_audio_chunk=False,
            return_tensors="pt",
        )
        inputs = inputs.to(model.device, dtype=model.dtype)
        yield inputs.input_features

        mel_frame_idx += processor.num_mel_frames_per_audio_chunk
        start_idx = mel_frame_idx * hop_length - n_fft // 2


# run generate in a thread, but print any error instead of failing silently
def run_generate(**kwargs):
    try:
        model.generate(**kwargs)
    except Exception:
        traceback.print_exc()


print(sd.query_devices(MIC, kind="input")["name"])
with sd.InputStream(
    device=MIC,
    samplerate=sampling_rate,
    channels=1,
    dtype="float32",
    blocksize=int(sampling_rate * 0.08),
    callback=mic_callback,
):
    print("Listening... speak now (press Interrupt to stop)")

    # create the first chunk only after recording enough audio
    wait_for_samples(processor.num_samples_first_audio_chunk)
    first_chunk_inputs = processor(
        buffer[: processor.num_samples_first_audio_chunk],
        sampling_rate=sampling_rate,
        is_streaming=True,
        is_first_audio_chunk=True,
        return_tensors="pt",
    )
    first_chunk_inputs = first_chunk_inputs.to(model.device, dtype=model.dtype)

    streamer = TextIteratorStreamer(processor.tokenizer, skip_special_tokens=True)
    generate_kwargs = {
        **first_chunk_inputs,
        "input_features": input_features_generator(first_chunk_inputs),
        "streamer": streamer,
        "max_new_tokens": 100_000,
    }
    thread = Thread(target=run_generate, kwargs=generate_kwargs, daemon=True)
    thread.start()

    print("Model output (streaming):", end=" ", flush=True)
    try:
        for text_chunk in streamer:
            print(text_chunk, end="", flush=True)
    except KeyboardInterrupt:
        stop.set()
        print("\nStopped.")
```
